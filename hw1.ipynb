idea:利用主定理来处理 $T(n) = aT(n/b) + f(n)$ 这一类的递归式。1.通过比较 $f(n)$ 与 $n^{\log_b a}$ 的增长速度来确定复杂度。2.如果 $f(n)$ 增长更慢，复杂度由递归树的叶子节点决定；如果 $f(n)$ 增长更快，复杂度由 $f(n)$ 决定；如果两者相当，则需额外乘以 $\log n$。3.对于非标准形式，则采用变量代换或递归树展开来观察规模缩减的速度。

1. $T(n) = T(n/4) + 1$。根据主定理 $T(n) = aT(n/b) + f(n)$。本题 $a=1, b=4, f(n)=n^0=1$。通过比较 $f(n)$ 与 $n^{\log_b a}$：$n^{\log_4 1} = n^0 = 1$。因为 $f(n) = \Theta(n^{\log_b a})$，根据主定理的第二种情况可知：$T(n) = O(n^{\log_b a} \cdot \log n) = \mathbf{O(\log n)}$。

2. $T(n) = 3T(n/3) + n^2$。根据主定理 $T(n) = aT(n/b) + f(n)$。本题 $a=3, b=3, f(n)=n^2$。通过比较 $f(n)$ 与 $n^{\log_b a}$：$n^{\log_3 3} = n^1$。因为 $f(n) = \Omega(n^{\log_b a + \epsilon})$（即 $n^2$ 多项式大于 $n^1$），根据主定理的第三种情况可知：$T(n) = \Theta(f(n)) = \mathbf{O(n^2)}$。

3. $T(n) = T(2n/3) + 1$。根据主定理 $T(n) = aT(n/b) + f(n)$。本题 $a=1, b=3/2, f(n)=n^0=1$。通过比较 $f(n)$ 与 $n^{\log_b a}$：$n^{\log_{1.5} 1} = n^0 = 1$。因为 $f(n) = \Theta(n^{\log_b a})$，根据主定理的第二种情况可知：$T(n) = O(n^{\log_b a} \cdot \log n) = \mathbf{O(\log n)}$。

4. $T(n) = 5T(n/4) + n$。根据主定理 $T(n) = aT(n/b) + f(n)$。本题 $a=5, b=4, f(n)=n^1$。通过比较 $f(n)$ 与 $n^{\log_b a}$：$n^{\log_4 5} \approx n^{1.16}$。因为 $f(n) = O(n^{\log_b a - \epsilon})$（即 $n^1$ 多项式小于 $n^{1.16}$），根据主定理的第一种情况可知：$T(n) = \Theta(n^{\log_b a}) = \mathbf{O(n^{\log_4 5})}$。

5. $T(n) = T(\sqrt{n}) + 1$。本题不直接适用主定理，需使用变量代换法。令 $n = 2^m$，则 $\sqrt{n} = 2^{m/2}$。代入原式得 $T(2^m) = T(2^{m/2}) + 1$。设 $S(m) = T(2^m)$，则 $S(m) = S(m/2) + 1$。对 $S(m)$ 使用主定理，符合第二种情况可知 $S(m) = O(\log m)$。将 $m = \log n$ 代回可得：$T(n) = \mathbf{O(\log \log n)}$

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea:本题的实质是在一支股票的波动序列中，寻找一段连续的时间区间，使得这段时间内收益的总和最大。因此可以将问题转化为求解 profits 数组的最大连续子数组的和，利用贪心算法的思想从头开始累加收益，并用一个变量记录过程中的最大值。如果当前累加收益跌为负数，说明之前的操作只会拖累后续收益，此时应将当前累加值清零，并从下一天起重新作为潜在的买入起点进行计算。并且该算法只需遍历数组一次，效率很高。

````python
# add your code here
def find_max_profit(profits):
    n = len(profits)    
    if n == 0:        
        return 0, None, None
        
    max_so_far = -float('inf')  
    current_max = 0               
    start = 0                   
    buy_date = 0                
    sell_date = 0                   
      
    for i in range(n):
        current_max += profits[i]                
        
        if current_max > max_so_far:
            max_so_far = current_max
            buy_date = start  
            sell_date = i + 1                   
                  
        if current_max < 0:
            current_max = 0
            start = i + 1                 
            
    return max_so_far, buy_date, sell_date    


profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]
max_p, buy, sell = find_max_profit(profits)

print(f"最大利润(收益最大): {max_p}")
print(f"买入日期: 第{buy+1}天")
print(f"卖出日期: 第{sell}天")
````

# your algorithm time complexity is: $O(n)$

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:本题可以利用二分法查找局部最小值即通过比较中间元素 $A[mid]$ 与其相邻元素的大小，从而判断当前区域处于“上坡”还是“下坡”。由于题目给定数组两端已知是一降一升(即两端高、中间低)，每次比较均可舍弃掉没有“谷底”的那一半区间，从而在 $O(\log n)$ 的时间复杂度内锁定其中一个局部最小值。

````python
# add your code here
def find_local_minimum(A):
    n = len(A)
    left, right = 1, n - 2
    
    while left <= right:
        mid = left + (right - left) // 2
        
        if A[mid] < A[mid - 1] and A[mid] < A[mid + 1]:
            return A[mid]
            
        elif A[mid] > A[mid - 1]:
            right = mid - 1
            
        else:
            left = mid + 1
            
    return None

A = [9, 3, 7, 2, 1, 4, 5]
local_min = find_local_minimum(A)
print(f"找到的局部最小值为: {local_min}")
````

# your algorithm time complexity is: $O(\log n)$

idea:利用堆这种数据结构的特性，初始化一个空的最小堆，遍历长度为 $n$ 的无序数组，将每个元素逐个执行 Push 插入操作，堆建立完成后，连续执行 $k$ 次提取堆顶的操作(Pop)，每次弹出的堆顶都是当前堆里的全局最小值，依次将它们记录下来，最后就能按升序输出前 $k$ 小的数值。

````python
# add your code here（第二问）
import heapq

def top_k_by_heap_insert(arr, k):
    min_heap = []
    
    for num in arr:
        heapq.heappush(min_heap, num)
        
    result = []
    for _ in range(k):
        min_element = heapq.heappop(min_heap) 
        result.append(min_element)
        
    return result

arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3
print("第二问的输出结果为:", top_k_by_heap_insert(arr, k))
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea:利用栈模拟星球的碰撞，遍历数组中的每个星球：仅当“栈顶星球向右(>0) 且 当前星球向左(<0)”时才会发生碰撞。碰撞时比较两者的绝对值，质量较小的被摧毁；若质量相等，则保留栈顶向右的星球。将最终幸存的星球入栈，遍历结束后栈内元素即为存活的星球。

````python
# add your code here
def planet_collisions(A):
    stack = []
    
    for p in A:
        alive = True 
        
        while alive and stack and stack[-1] > 0 and p < 0:
            if stack[-1] < -p:
                stack.pop()
                
            elif stack[-1] == -p:
                alive = False
                
            else:
                alive = False
                
        if alive:
            stack.append(p)
            
    return stack

print("测试 1:", planet_collisions([-3, -6, 2, 8, 5, -8, 9, -2, 1]))
print("测试 2:", planet_collisions([23, -8, 9, -3, -7, 9, -23, 22]))
````

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:利用动态规划算法的思想用dp数组去记录推车在各种承重下能装的最大价值。然后把4块石头分别拿过来试。每拿到一块石头，就根据推车的最大承重开始倒着往回计算，如果推车还能装下这块石头，就考虑把它装进去好，还是保持原样更好。这样试完所有的大理石，就能得到最大承重对应的价值。

````python
# add your code here
def knapsack_01(items, max_capacity):
    dp = [0] * (max_capacity + 1)
    
    for weight, value in items:
        for j in range(max_capacity, weight - 1, -1):
            dp[j] = max(dp[j], dp[j - weight] + value)
            
    return dp[max_capacity]

marbles = [(5, 10), (4, 40), (6, 30), (3, 50)]
capacity = 9

max_value = knapsack_01(marbles, capacity)
print(f"小推车能装载的最大价值为: {max_value}")
````

2.

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:利用二叉搜索树的中序遍历必然是严格递增序列的特性，在遍历时发现前面的数比后面的大，就说明节点被换错了。如果是相邻节点换错，则降序只会发生一次，但如果是不相邻节点换错，那么降序会发生两次。因此只需要在遍历中找到这两个导致降序的节点，把它们的值对调回来即可。

````python
# add your code here
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def recoverTree(root):
    first = second = prev = None
    
    def inorder(node):
        nonlocal first, second, prev
        if not node:
            return
        
        inorder(node.left)
        
        if prev and prev.val > node.val:
            if not first:
                first = prev
            second = node
        prev = node
        
        inorder(node.right)
        
    inorder(root)
    
    if first and second:
        first.val, second.val = second.val, first.val


def build_tree(lst):
    if not lst: return None
    root = TreeNode(lst[0])
    queue = deque([root])
    i = 1
    while queue and i < len(lst):
        node = queue.popleft()
        if lst[i] is not None:
            node.left = TreeNode(lst[i])
            queue.append(node.left)
        i += 1
        if i < len(lst) and lst[i] is not None:
            node.right = TreeNode(lst[i])
            queue.append(node.right)
        i += 1
    return root

def tree_to_list(root):
    if not root: return []
    result = []
    queue = deque([root])
    while queue:
        node = queue.popleft()
        if node:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            result.append(None)
    while result and result[-1] is None:
        result.pop()
    return result

input_1 = [1, 3, None, None, 2]
root1 = build_tree(input_1)
in_1_str = str(input_1).replace('None', 'null')
print(f"示例 1 输入: root = {in_1_str}")
recoverTree(root1)
out_1 = str(tree_to_list(root1)).replace('None', 'null')
print(f"输出为: {out_1}")  


input_2 = [3, 1, 4, None, None, 2]
root2 = build_tree(input_2)
in_2_str = str(input_2).replace('None', 'null')
print(f"示例 2 输入: root = {in_2_str}")
recoverTree(root2)
out_2 = str(tree_to_list(root2)).replace('None', 'null')
print(f"输出为: {out_2}")
````

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:利用深度优先搜索自顶向下遍历二叉树。在往下走的过程中，像进位一样，把父节点传下来的数字乘以10，再加上当前节点的值。如果走到没有子节点的叶节点，就说明找到了一条完整的路径，就把这条路径代表的数字返回。最后把所有叶节点返回的数字累加起来即可。

````python
# add your code here
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def sumNumbers(root):
    def dfs(node, current_sum):
        if not node:
            return 0
        current_sum = current_sum * 10 + node.val
        if not node.left and not node.right:
            return current_sum
        return dfs(node.left, current_sum) + dfs(node.right, current_sum)
    return dfs(root, 0)

def build_tree(lst):
    if not lst: return None
    root = TreeNode(lst[0])
    queue = deque([root])
    i = 1
    while queue and i < len(lst):
        node = queue.popleft()
        if lst[i] is not None:
            node.left = TreeNode(lst[i])
            queue.append(node.left)
        i += 1
        if i < len(lst) and lst[i] is not None:
            node.right = TreeNode(lst[i])
            queue.append(node.right)
        i += 1
    return root

def format_list(lst):
    return str(lst).replace('None', 'null')


input_1 = [1, 2, 3]
root1 = build_tree(input_1)
print(f"例 1 输入: root = {format_list(input_1)}")
print(f"输出为: {sumNumbers(root1)}")  


input_2 = [4, 9, 0, 5, 1]
root2 = build_tree(input_2)
print(f"例 2 输入: root = {format_list(input_2)}")
print(f"输出为: {sumNumbers(root2)}")
````

# your algorithm time complexity is: $O(n)$
